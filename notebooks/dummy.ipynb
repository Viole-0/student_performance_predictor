```python
import pandas as pd

# Load dataset with correct delimiter
df = pd.read_csv("student-mat.csv", delimiter=";")

# Drop the target column (G3) and encode categoricals
df_encoded = pd.get_dummies(df.drop(columns=["G3"]), drop_first=True)

# Save the list of input features
import joblib
joblib.dump(df_encoded.columns.tolist(), "trained_feature_names.pkl")
```
